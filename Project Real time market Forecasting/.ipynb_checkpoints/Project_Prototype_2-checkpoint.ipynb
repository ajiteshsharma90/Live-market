```python
import yfinance as yf
import numpy as np
import pandas as pd
import tensorflow as tf
import matplotlib.pyplot as plt
from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import mean_absolute_error, mean_squared_error, accuracy_score
from tqdm import tqdm

# Configuration
TICKER = "AAPL"  # Replace with your stock symbol
LOOKBACK = 60    # Number of past days to use for prediction
FORECAST_DAYS = 10  # Number of days to predict into the future
TEST_SIZE = 30   # Number of days to use for testing
EPOCHS = 50      # Number of training epochs

# Fetch Data
def fetch_data(ticker, period="2y", interval="1d"):
    """Fetch historical data for a single stock"""
    data = yf.download(ticker, period=period, interval=interval, progress=False)
    return data

# Preprocess Data
def preprocess_data(data, lookback):
    """Prepare data for LSTM model"""
    scaler = MinMaxScaler(feature_range=(0, 1))
    scaled_data = scaler.fit_transform(data[['Close', 'Volume']].values)

    # Create sequences
    X, y = [], []
    for i in range(lookback, len(scaled_data)):
        X.append(scaled_data[i-lookback:i])
        y.append(scaled_data[i, 0])  # Predict Close price
    return np.array(X), np.array(y), scaler

# Build LSTM Model
def build_lstm_model(input_shape):
    """Create an LSTM model"""
    model = tf.keras.Sequential([
        tf.keras.layers.LSTM(128, return_sequences=True, input_shape=input_shape),
        tf.keras.layers.Dropout(0.2),
        tf.keras.layers.LSTM(64),
        tf.keras.layers.Dense(1)
    ])
    model.compile(optimizer='adam', loss='mse')
    return model

# Train Model
def train_model(model, X_train, y_train, epochs=50):
    """Train the LSTM model"""
    model.fit(X_train, y_train, epochs=epochs, batch_size=32, verbose=1)
    return model

# Generate Predictions
def generate_predictions(model, data, scaler, lookback, test_size):
    """Generate predictions for the test set"""
    predictions = []
    actuals = []

    # Prepare test data
    test_data = data[-test_size - lookback:]
    scaled_test_data = scaler.transform(test_data[['Close', 'Volume']])

    # Walk-forward validation
    for i in range(lookback, len(scaled_test_data)):
        sequence = scaled_test_data[i-lookback:i]
        pred = model.predict(sequence.reshape(1, lookback, 2))
        predictions.append(pred[0, 0])
        actuals.append(scaled_test_data[i, 0])

    # Inverse transform predictions
    predictions = np.array(predictions).reshape(-1, 1)
    actuals = np.array(actuals).reshape(-1, 1)

    # Create dummy volume data for inverse transform
    dummy_vol = np.zeros_like(predictions)
    predictions = scaler.inverse_transform(np.concatenate([predictions, dummy_vol], axis=1))[:, 0]
    actuals = scaler.inverse_transform(np.concatenate([actuals, dummy_vol], axis=1))[:, 0]

    return actuals, predictions

# Calculate Directional Accuracy
def calculate_directional_accuracy(actuals, predictions):
    """Calculate directional accuracy (up/down predictions)"""
    actual_direction = np.diff(actuals) > 0  # True if price went up
    predicted_direction = np.diff(predictions) > 0  # True if predicted to go up
    return accuracy_score(actual_direction, predicted_direction)

# Plot Results
def plot_actual_vs_predicted(actuals, predictions, ticker):
    """Plot actual vs predicted prices"""
    plt.figure(figsize=(14, 7))
    plt.plot(actuals, label='Actual Prices', marker='o')
    plt.plot(predictions, label='Predicted Prices', marker='x')
    plt.title(f'{ticker} Actual vs Predicted Prices')
    plt.xlabel('Days')
    plt.ylabel('Price ($)')
    plt.legend()
    plt.grid(True)
    plt.show()

# Main Function
def main():
    # Step 1: Fetch Data
    data = fetch_data(TICKER)

    # Step 2: Preprocess Data
    X, y, scaler = preprocess_data(data, LOOKBACK)

    # Split into train/test sets
    train_size = len(X) - TEST_SIZE
    X_train, y_train = X[:train_size], y[:train_size]
    X_test, y_test = X[train_size:], y[train_size:]

    # Step 3: Build and Train Model
    model = build_lstm_model((X_train.shape[1], X_train.shape[2]))
    model = train_model(model, X_train, y_train, epochs=EPOCHS)

    # Step 4: Generate Predictions
    actuals, predictions = generate_predictions(model, data, scaler, LOOKBACK, TEST_SIZE)

    # Step 5: Calculate Directional Accuracy
    directional_accuracy = calculate_directional_accuracy(actuals, predictions)

    # Step 6: Plot Results
    plot_actual_vs_predicted(actuals, predictions, TICKER)

    # Step 7: Calculate Metrics
    mae = mean_absolute_error(actuals, predictions)
    rmse = np.sqrt(mean_squared_error(actuals, predictions))
    mape = np.mean(np.abs((actuals - predictions) / actuals)) * 100
    print(f"\nMetrics for {TICKER}:")
    print(f"MAE: {mae:.2f}")
    print(f"RMSE: {rmse:.2f}")
    print(f"MAPE: {mape:.2f}%")
    print(f"Directional Accuracy: {directional_accuracy * 100:.2f}%")

if __name__ == "__main__":
    main()
```
